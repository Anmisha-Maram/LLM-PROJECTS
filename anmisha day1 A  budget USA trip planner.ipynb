2.Make the messages list

```python
messages = [
    {"role": "system", "content": "You are a budget guide assistant"},
    {"role": "user", "content": "What are the attractions in cincinnati,ohio?"}
]

```

3. Call openAPI

```python
openai = OpenAI()

response = openai.chat.completions.create(model="gpt-5-nano", messages=messages)
response.choices[0].message.content
```
