```python
# imports
import os
import ollama
from openai import OpenAI
from dotenv import load_dotenv


load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')
#client = OpenAI()  # uses OPENAI_API_KEY from env

if not api_key:
    print("no api key found- please check your .env file")
elif not api_key:
    print("API key was found, but it is not starting with sk-proj")
else:
    print("API key found and looks good")
```

# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'

```python
# set up environment
MODEL_GPT = "gpt-4o-mini"
MODEL_LLAMA = "llama3.2"
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming
openai = OpenAI()
response = openai.chat.completions.create(
    model=MODEL_GPT,
    messages=[
        {"role": "system", "content": "You are a smart, clever, expert technical explainer. Provide a concise, structured explanation with key points, examples and why it matters. Use bullet points."},
        {"role": "user", "content": question}
    ],
)
response.choices[0].message.content
```

```python
print(response.choices[0].message.content)
```

```python
OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')
```

```python

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get Llama 3.2 to answer
response = ollama.chat.completions.create(
    model= MODEL_LLAMA,
    messages = [
        {"role": "system", "content": " You are an expert technical explainer. Provide a concise, structed explanation with key points, use bullet points."},
        {"role": "user", "content": question}
    ]
)

response.choices[0].message.content
```
